```python
import mne
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
import os
from Parameters import path_directory, subject_to_pop
from Import_data import import_df_description, Make_path_list
from Tools import  find_participant
import seaborn as sns
import pandas as pd
```

```python
path_data, path_stim, path_answers = Make_path_list(path_directory,subject_to_pop)

list_subject = []
for sub_path in path_data:
    list_subject.append(find_participant(sub_path,signe = 'S'))
```

```python
liste_individual = []
liste_average = []

for sub in list_subject:

    group_average = mne.read_evokeds(f"../group_average_{sub}.fif")[0]
    subject_average = mne.read_evokeds(f"../subject_average_{sub}.fif")[0]
    liste_individual.append(subject_average.data)
    liste_average.append(group_average.data)
```

```python
def compute_RMSE(sub,group):
    return np.sqrt(np.mean((sub - group) ** 2))
```

```python
liste_RMSE = []
for sub, group in zip(liste_individual,liste_average):
    liste_RMSE.append(compute_RMSE(sub,group))


array_sub = np.array(list_subject)
array_RMSE = np.array(liste_RMSE) * 10e6

dico = {
    "Sub" : array_sub,
    "RMSE" : array_RMSE ,
}

df = pd.DataFrame(dico)
```

```python
meta_df = pd.read_csv("..\meta_data_subjects.csv")
meta_df["rmse_inra"]
df["err_RMSE"] = meta_df["rmse_inra"] * 100
```

```python
df = df.sort_values("RMSE")
# plt.ylim(25, 40)
sns.set(rc={"figure.figsize": (15, 8)})

plt.title("RMSE model moyenne")
sns.barplot(data = df, y = "RMSE", x = "Sub")


plt.errorbar(
    df["Sub"],
    df["RMSE"],
    yerr=df["err_RMSE"],   
    fmt='none',
    ecolor='black',
    capsize=5
)

plt.show()


print("===== stats RMSE du model paralelle =====")
print("mean = ",array_RMSE.mean())
print("std = ",array_RMSE.std())
print("max = ",array_RMSE.max())
print("min = ",array_RMSE.min())
print("=========================================")
```
