```python
import numpy as np
from osl_dynamics.models.hmm import Model, Config
```

```python
# Fake EEG data (time x channels)
data = np.random.randn(10000, 10)

# Config du modèle
config = Config(
    n_states = 3,
    n_channels = 10,
    sequence_length = 200,
    batch_size=32 ,
    learn_means= False,
    learn_covariances= True,
    n_epochs=10,
    learning_rate=0.001,
)
```

```python
# Création du modèle
model = Model(config)
```

```python

# Entraînement
model.fit(data)
```

```python

# Inférence des états
alpha = model.get_alpha(data)
```

```python

# États les plus probables
states = np.argmax(alpha, axis=1)

# for el in states:
#     print(el)
len(states)
```
