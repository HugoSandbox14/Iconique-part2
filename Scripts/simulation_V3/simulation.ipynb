```python
path_data, path_stim, path_answers = Make_path_list(path_directory,subject_to_pop)

list_subject = []
for sub_path in path_data:
    list_subject.append(find_participant(sub_path,signe = 'S'))
```

```python
list_true_eeg = []
for sub in list_subject:

    group_average = mne.read_evokeds(f"../group_average_{sub}.fif")[0]
    subject_epochs = mne.read_epochs(f"../subject_epochs_{sub}.fif")
    list_true_eeg.append(True_EEG(subject_epochs,group_average,subject = sub))

```

```python
base = Requirement_Simulation()
```

```python
count = 0
for el in list_true_eeg:
   count+= el.epochs_individual.get_data().shape[0]
print(count)                                          # nombre total d'epoch etudie
```

```python
a = list_true_eeg[0]                        
print(count - a.average_group.nave)         # nombre d'epoch total - nombre d'epoch dans le groupe utilise pour la simulation
print(len(a.epochs_individual))             # nombre d'epoch du sujet 1
```

```python
list_simulations_para = []
list_simulations_seri = []
list_model_moyenne = []

for eeg in list_true_eeg:
    list_simulations_para.append(Simulation_Parallele(base,eeg,1e-10,methode="pinv"))
    list_simulations_seri.append(Simulation_Serielle(base,eeg,1e-10,methode="pinv"))
    list_model_moyenne.append(Model_Moyenne(eeg))
```

```python
group_simulation_para = Group_simulation(list_simulations_para, model_name="Paralel", from_average=True)
group_simulation_seri = Group_simulation(list_simulations_seri, model_name="Seriel", from_average=True)
group_model_moyenne = Group_simulation(list_model_moyenne, model_name="Moyenne", from_average=True)

group_simulation_para.plot(sort = "mean",info_type="std",value = "rmse")
# group_simulation_para.plot(sort = "mean",info_type="std",value = "corr")
group_simulation_para.display_main_info()

group_simulation_seri.plot(sort = "mean",info_type="std",value = "rmse")
# group_simulation_seri.plot(sort = "mean",info_type="std",value = "corr")
group_simulation_seri.display_main_info()

group_model_moyenne.plot(sort = "mean",info_type="std",value = "rmse")
# group_model_moyenne.plot(sort = "mean",info_type="std",value = "corr")


group_model_moyenne.display_main_info()

```

```python
time = np.linspace(-0.2,1.2,sim_1_para.Alpha_matrix.shape[1])
labels = ["VisuoC",'LexicoS','LexicoP','Phonologique','Articulation','Bruit']

print("Paralelle1")
for region, l in zip(sim_1_para.Alpha_matrix,labels):
    plt.plot(time,region,label = l)
    plt.title("Source du model paralel")
plt.legend()
plt.show()

print("Seriel")
for region, l in zip(sim_1_seri.Alpha_matrix,labels):
    plt.plot(time,region,label = l)
    plt.title("Source du model seriel")
plt.legend()
plt.show()
```
